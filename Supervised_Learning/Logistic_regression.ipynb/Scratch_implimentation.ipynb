```python

import numpy as np

class logisticRegression:

    def __init__(self, learning_rate=0.01, n_iter=1000):
        self.learning_rate = learning_rate
        self.n_iter = n_iter
        self.weights = None
        self.bias = None

    def _sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    def fit(self, X, y):

        n_samples, n_features = X.shape

        self.weights = np.zeros(n_features)
        self.bias = 0

        for _ in range(self.n_iter):

            # Forward propagation
            linear_model = self.bias + np.dot(X, self.weights)

            y_predicted = self._sigmoid(linear_model)

            # Calculate gradients
            dw = (1 / n_samples) * np.dot(
                X.T, (y_predicted - y)
            )

            db = (1 / n_samples) * np.sum(
                y_predicted - y
            )

            # Update weights and bias
            self.weights -= self.learning_rate * dw
            self.bias -= self.learning_rate * db

    def get_probabilities(self, X):

        linear_model = self.bias + np.dot(X, self.weights)

        return self._sigmoid(linear_model)

    def predict(self, X):

        y_predicted = self.get_probabilities(X)

        return np.array([
            1 if i > 0.5 else 0 for i in y_predicted
        ])


# Dataset
X = np.array([
    [1,2],
    [2,3],
    [3,4],
    [4,5]
])

Y = np.array([0,0,0,1])

# Model
model = logisticRegression()

model.fit(X, Y)

print("Predictions:", model.predict(X))
print("Probabilities:", model.get_probabilities(X))
print("Weights:", model.weights)
print("Bias:", model.bias)
```
